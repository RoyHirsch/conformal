```python
import numpy as np

# log_file = '/home/royhirsch/conformal/exps/organamnist/organamnist_alpha=0.1_0211/net_launcher_log.log'
# log_file = '/home/royhirsch/conformal/exps/organamnist/organamnist_alpha=0.05_0211/net_launcher_log.log'

# log_file = '/home/royhirsch/conformal/exps/tissuemnist/tissuemnist_alpha=0.1_0211/net_launcher_log.log'
log_file = '/home/royhirsch/conformal/exps/tissuemnist/tissuemnist_alpha=0.05_0211/net_launcher_log.log'

with open(log_file, 'r') as f:
    data = f.readlines()

t = []
for line in data:
    if 'Residuals correction' in line:
        t.append(line)


ls = []
hs = []
for line in t:
    l, h = line.split('[')[1].split(']')[0].split(', ')
    ls.append(l)
    hs.append(h)

ls = np.asarray([float(i) for i in ls])
hs = np.asarray([float(i) for i in hs])
print('Low qhat: {:.3f} ({:.3f})'.format(np.mean(ls), np.std(ls)))
print('High qhat: {:.3f} ({:.3f})'.format(np.mean(hs), np.std(hs)))
```
